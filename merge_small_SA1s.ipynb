```python
import geopandas as gpd
import pandas as pd
pd.set_option('display.max_columns', None)
```

```python
sa1 = gpd.read_file("input/statsnz-2023-census-totals-by-topic-for-individuals-by-statistical-a-SHP.zip")
sa1
```

```python
sa1["SA12023_V1"] = sa1["SA12023_V1"].astype(int)
```

```python
townsend = pd.read_excel("input/townsend.xlsx")
townsend
```

```python
sa1 = sa1[["SA12023_V1", "geometry"]].merge(townsend, left_on="SA12023_V1", right_on="sa1_2023").sort_values("urpop_2023")
sa1
```

```python
dz = sa1[sa1.SA12023_V1 == 7029838]
dz
```

```python
neighbours = sa1[sa1.geometry.touches(dz.geometry.iloc[0])]
neighbours
```

```python
ax = neighbours.boundary.plot()
ax = dz.boundary.plot(ax=ax, color="red", linewidth=2)
p = dz.centroid.iloc[0]
ax.annotate(text=str(dz.urpop_2023.iloc[0]), xy=[p.x, p.y], ha='center', color="blue")
for neighbour in neighbours.itertuples():
  p = neighbour.geometry.centroid
  ax.annotate(text=str(neighbour.urpop_2023), xy=[p.x, p.y], ha='center', color="blue")
```

```python
neighbours["tds_diff"] = abs(neighbours.tds - dz.tds.iloc[0])
neighbours.sort_values("tds_diff", inplace=True)
neighbours
```

```python
merged = pd.concat((dz, neighbours.iloc[0:1])).dissolve(aggfunc={
    "sa1_2023": "first",
    "urpop_2023": "sum",
    "tds": "mean"
})
merged.index = dz.index
merged
```

```python
ax = neighbours.boundary.plot()
merged.boundary.plot(ax=ax, color="red")
```

```python
while sa1.urpop_2023.iloc[0] < 399:
    dz = sa1.iloc[0:1]
    neighbours = sa1[sa1.geometry.touches(dz.geometry.iloc[0])].copy()
    if len(neighbours) == 0:
        print(f"No neighbours found for {dz.sa1_2023.iloc[0]}")
        sa1.drop(dz.index[0], inplace=True)
        continue
    neighbours["tds_diff"] = abs(neighbours.tds - dz.tds.iloc[0])
    neighbours.sort_values("tds_diff", inplace=True)
    merged = pd.concat((dz, neighbours.iloc[0:1])).dissolve(aggfunc={
        "sa1_2023": "first",
        "urpop_2023": "sum",
        "tds": "mean"
    })
    print(f"Merging {dz.sa1_2023.iloc[0]} ({dz.urpop_2023.iloc[0]}) with {neighbours.sa1_2023.iloc[0]} ({neighbours.urpop_2023.iloc[0]}) into {merged.sa1_2023.iloc[0]} ({merged.urpop_2023.iloc[0]})")
    merged.index = dz.index
    sa1.drop(neighbours.index[0], inplace=True)
    sa1.update(merged)
    sa1.sort_values("urpop_2023", inplace=True)
sa1
```

```python
sa1.to_file("output/sa1_2023_merged_small_dz.shp.zip")
```
